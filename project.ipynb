```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = pd.read_csv(r'D:\AI\premier_league_practice_dataset.csv')
print(df.head())
```

```python
arsenal_data = df[df['Team'] == 'Arsenal']
plt.figure(figsize=(10, 5))
sns.lineplot(x='Season',y='Points',data=arsenal_data,marker='o',color='red',linewidth=2.5,)
plt.title('Points Earned by Arsenal Across Seasons', fontsize=14)
plt.xlabel('Season', fontsize=12)
plt.ylabel('Points', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Observation:
# Arsenal's performance in terms of points shows fluctuations over the seasons,
# highlighting variations in their overall placement and performance.
```

```python

teams = ['Arsenal', 'Liverpool', 'Manchester City']
filtered_df = df[df['Team'].isin(teams)]

plt.figure(figsize=(12, 6))
sns.lineplot(data=filtered_df, x='Season',y='Points', hue='Team',marker='o',linewidth=2)
plt.title('Points Comparison: Arsenal vs Liverpool vs Manchester City', fontsize=14)
plt.xlabel('Season', fontsize=12)
plt.ylabel('Points', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Observation:
# Manchester City and Liverpool consistently show higher point totals in recent seasons compared to Arsenal, 
# with Manchester City maintaining the most consistent top-tier performance across the dataset.
```

```python
tottenham_data = df[df['Team'] == 'Tottenham']
plt.figure(figsize=(10, 5))
sns.lineplot(x='Season', y='Goals_For', data=tottenham_data, marker='o', color='blue', linewidth=2.5)
plt.title('Goals Scored by Tottenham Across Seasons', fontsize=14)
plt.xlabel('Season', fontsize=12)
plt.ylabel('Goals For', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Observation: Tottenham's goals scored show fluctuations across seasons, reflecting variations in their attacking performance.
```

```python
filtered_df = df[df['Team'].isin(['Chelsea', 'Liverpool', 'Manchester City'])]
plt.figure(figsize=(12, 6))
sns.lineplot(data=filtered_df, x='Season', y='Possession_Percent', hue='Team', marker='o', linewidth=2)
plt.title('Possession Percentage Comparison: Chelsea, Liverpool, and Manchester City Across Seasons', fontsize=14)
plt.xlabel('Season', fontsize=12)
plt.ylabel('Possession Percent', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Observation: Manchester City consistently maintains the highest possession percentage compared to Chelsea and Liverpool across the seasons.
```
